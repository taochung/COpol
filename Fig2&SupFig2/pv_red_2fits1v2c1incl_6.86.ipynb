```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, FormatStrFormatter
from astropy.io import fits
import matplotlib.patches as patches
import matplotlib as mpl
font = {'size'   : 32}
mpl.rc('font', **font)
minorLocator = MultipleLocator(10)

fig = plt.figure(figsize = (32,32))
fig.subplots_adjust(0.25,0.25,0.85,0.85,0.15,0.15)
ax0 = fig.add_subplot(1,1,1)
ax0.set_xticklabels([])
ax0.set_yticklabels([])
ax0.axis('off')

vlsr = 6.86

v5=(38.525+3.175*8)*1000
v6=(38.525+3.175*7)*1000
v7=(38.525+3.175*6)*1000
v8=(38.525+3.175*5)*1000
v9=(38.525+3.175*4)*1000
v10=(38.525+3.175*3)*1000
v11=(38.525+3.175*2)*1000
v12=(38.525+3.175*1)*1000
v13=(38.525*1000)
v14=(35.350*1000)
v15=(32.175*1000)
v16=(29.000*1000)
v17=(25.825*1000)
v18=(22.650*1000)
v19=(19.475*1000)
v20=(16.300*1000)
v21=(13.125*1000)
v22= (9.950*1000)
v23= (9.950-3.175)*1000
v24= (9.950-3.175*1)*1000
v25= (9.950-3.175*2)*1000
v26= (9.950-3.175*3)*1000
v27= (9.950-3.175*4)*1000
v28= (9.950-3.175*5)*1000
v29= (9.950-3.175*6)*1000
v30= (9.950-3.175*7)*1000

poff_4A1 = 1.0045 #1.0045 arcsec from crval2

#################
#####major pv####
#################
ax1 = fig.add_subplot(2,2,1)
d1 = fits.open("/Users/tching/Desktop/proposals/ALMA/Cycle4/COpolarization/python/sn3_10/data/moment/red.i.pv_pa171.5.fits")
cdelt1 = d1[0].header['CDELT1'] #delta v
crval1 = d1[0].header['CRVAL1'] #+cdelt1/2 #v0
cdelt2 = d1[0].header['CDELT2'] #delta p
crval2 = d1[0].header['CRVAL2'] #p0
h0 = d1[0].header
#print("cdelt1=",cdelt1)
#print("crval1=",crval1)

data1 = d1[0].data
pv = data1.squeeze()
pvM_rot = pv #np.flipud(pv)
d1.close()
p_max = 10
p_min = -10
v_min = -30*1000
v_max = 70*1000
y_min = (p_min-crval2)/cdelt2
y_max = (p_max-crval2)/cdelt2
x_min = (v_min-crval1)/cdelt1
x_max = (v_max-crval1)/cdelt1
#print(x_min,x_max,y_min,y_max)

######################
### model pv curve ###
######################
r = np.linspace(-10, 10, 100)
v0 = 0.08
const = 89.8
incl = 89*np.pi/180 #20*np.pi/180
pv_v = v0*r*(np.cos(incl) + const*r*np.sin(incl))
pv_p = np.cos(incl)*const*r**2 - r*np.sin(incl)
pv_v_red1 = ((vlsr + pv_v)*1000-crval1)/cdelt1
pv_p_red1 = (-pv_p - crval2)/cdelt2 + poff_4A1/cdelt2
pv_v_blue1 = ((vlsr - pv_v)*1000-crval1)/cdelt1
pv_p_blue1 = (pv_p - crval2)/cdelt2 + poff_4A1/cdelt2
#ax1.plot(pv_v_red1, pv_p_red1,lw=0.4, linestyle='-', color='r')
#ax1.plot(pv_v_blue1, pv_p_blue1,lw=0.4, linestyle='-', color='r')

ax1.plot([x_min,x_max], [(poff_4A1-crval2)/cdelt2,(poff_4A1-crval2)/cdelt2],lw=0.4, linestyle=':', color='k')
ax1.plot([(vlsr*1000-crval1)/cdelt1,(vlsr*1000-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='k')
ax1.plot([(v15-crval1)/cdelt1,(v15-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='b')
ax1.plot([(v16-crval1)/cdelt1,(v16-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='b')
ax1.plot([(v17-crval1)/cdelt1,(v17-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='b')
ax1.plot([(v18-crval1)/cdelt1,(v18-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='b')
ax1.plot([(v19-crval1)/cdelt1,(v19-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='b')
ax1.plot([(v20-crval1)/cdelt1,(v20-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='b')
ax1.plot([(v21-crval1)/cdelt1,(v21-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='b')
ax1.plot([(v22-crval1)/cdelt1,(v22-crval1)/cdelt1],[y_min,y_max], lw=0.4, linestyle=':', color='b')

ax1.scatter((v29-crval1)/cdelt1,np.argmax(pvM_rot[:,29]), 600, c='r', marker="+", zorder=50)
ax1.scatter((v27-crval1)/cdelt1,np.argmax(pvM_rot[:,27])+40, 600, c='r', marker="+", zorder=50)
ax1.scatter((v21-crval1)/cdelt1,np.argmax(pvM_rot[:,21]), 600, c='r', marker="+", zorder=50)
ax1.scatter((v20-crval1)/cdelt1,np.argmax(pvM_rot[:,20]), 600, c='r', marker="+", zorder=50)
ax1.scatter((v19-crval1)/cdelt1,np.argmax(pvM_rot[:,19]), 600, c='r', marker="+", zorder=50)
ax1.scatter((v18-crval1)/cdelt1,np.argmax(pvM_rot[:,18]), 600, c='c', marker="+", zorder=50)
ax1.scatter((v17-crval1)/cdelt1,np.argmax(pvM_rot[:,17]), 600, c='c', marker="+", zorder=50)
ax1.scatter((v18-crval1)/cdelt1,np.argmax(pvM_rot[100:200,18])+100, 600, c='r', marker="+", zorder=50)
ax1.scatter((v17-crval1)/cdelt1,np.argmax(pvM_rot[100:200,17])+100, 600, c='r', marker="+", zorder=50)
ax1.scatter((v17-crval1)/cdelt1,np.argmax(pvM_rot[200:250,17])+200, 600, c='r', marker="+", zorder=50)

ax1.scatter((v22-crval1)/cdelt1,np.argmax(pvM_rot[:,22]), 600, c='c', marker="+", zorder=50)
ax1.scatter((v16-crval1)/cdelt1,np.argmax(pvM_rot[:,16]), 600, c='c', marker="+", zorder=50)
ax1.scatter((v15-crval1)/cdelt1,np.argmax(pvM_rot[:,15]), 600, c='c', marker="+", zorder=50)
ax1.scatter((v27-crval1)/cdelt1,np.argmax(pvM_rot[:285,27]), 600, c='c', marker="+", zorder=50)
#ax1.scatter((v26-crval1)/cdelt1,np.argmax(pvM_rot[:285,26]), 600, c='c', marker="+", zorder=50)
#ax1.scatter((v25-crval1)/cdelt1,np.argmax(pvM_rot[:285,25]), 600, c='c', marker="+", zorder=50)
#ax1.scatter((v6-crval1)/cdelt1,np.argmax(pvM_rot[:,6]), 600, c='c', marker="+", zorder=50)
ax1.scatter((v12-crval1)/cdelt1,np.argmax(pvM_rot[:,12]), 600, c='c', marker="+", zorder=50)

#ax1.scatter((v29-crval1)/cdelt1+2*(vlsr*1000-v29)/cdelt1, \
#            2*(poff_4A1-crval2)/cdelt2-np.argmax(pvM_rot[:,29]), 15, c='y', marker="+", zorder=50)
#ax1.scatter((v27-crval1)/cdelt1+2*(vlsr*1000-v27)/cdelt1, \
#            2*(poff_4A1-crval2)/cdelt2-(np.argmax(pvM_rot[:,27])+40), 15, c='y', marker="+", zorder=50)

#print((v29-crval1)/cdelt1+2*(vlsr*1000-v29)/cdelt1,2*(poff_4A1-crval2)/cdelt2-np.argmax(pvM_rot[:,29]))
#print((v27-crval1)/cdelt1+2*(vlsr*1000-v27)/cdelt1,2*(poff_4A1-crval2)/cdelt2-(np.argmax(pvM_rot[:,27])+40))

#######
def wind(vinput,v0,const1,const2,incl):
    v = np.abs(vinput)
    num = np.where(vinput > 0 , 0, 1)
    coeff2_1 = np.zeros(len(y_input1))+v0*const1*np.sin(incl)
    coeff2_2 = np.zeros(len(y_input2))+v0*const2*np.sin(incl)
    coeff1_1 = np.zeros(len(y_input1))+v0*np.cos(incl)
    coeff1_2 = np.zeros(len(y_input2))+v0*np.cos(incl)
    coeff0_1 = -1*v[:len(y_input1)]
    coeff0_2 = -1*v[len(y_input1):]
    sol_1 = [-coeff1_1-np.sqrt(coeff1_1**2-4*coeff2_1*coeff0_1),-coeff1_1+np.sqrt(coeff1_1**2-4*coeff2_1*coeff0_1)]/(2*coeff2_1)
    sol_2 = [-coeff1_2-np.sqrt(coeff1_2**2-4*coeff2_2*coeff0_2),-coeff1_2+np.sqrt(coeff1_2**2-4*coeff2_2*coeff0_2)]/(2*coeff2_2)
    p1_1 = np.cos(incl)*const1*sol_1**2 - sol_1*np.sin(incl)
    p1_2 = np.cos(incl)*const2*sol_2**2 - sol_2*np.sin(incl)
    p1 = np.concatenate((p1_1,p1_2),axis=1)
    return (-p1[num,np.arange(0,len(vinput))] - crval2)/cdelt2 + poff_4A1/cdelt2

###### fit1 ######
v0 = 10.
const1 = 10.
const2 = 1.
incl = 15*np.pi/180 #20*np.pi/180
x_input1 = np.array([v21/1000-vlsr, v20/1000-vlsr, v18/1000-vlsr, v17/1000-vlsr,-(v17/1000-vlsr), \
                   -(vlsr-v29/1000), vlsr-v27/1000, v19/1000-vlsr])
y_input1 = np.array([np.argmax(pvM_rot[:,21]),np.argmax(pvM_rot[:,20]),np.argmax(pvM_rot[100:200,18])+100, \
                    np.argmax(pvM_rot[100:200,17])+100,np.argmax(pvM_rot[200:250,17])+200, 246.5387, 224.538721055118, np.argmax(pvM_rot[:,19])])
x_input2 = np.array([v22/1000-vlsr, v27/1000-vlsr, -(v12/1000-vlsr)])
y_input2 = np.array([np.argmax(pvM_rot[:,22]),np.argmax(pvM_rot[:285,27]),np.argmax(pvM_rot[:,12])])
#x_input2 = np.array([v22/1000-vlsr, -(v16/1000-vlsr), -(v15/1000-vlsr), v27/1000-vlsr, -(v18/1000-vlsr), \
#                   -(v17/1000-vlsr), -(v12/1000-vlsr)])
#y_input2 = np.array([np.argmax(pvM_rot[:,22]),np.argmax(pvM_rot[:,16]),np.argmax(pvM_rot[:,15]), \
#                    np.argmax(pvM_rot[:285,27]),np.argmax(pvM_rot[:,18]), np.argmax(pvM_rot[:,17]), np.argmax(pvM_rot[:,12])])
x0 = [v0,const1,const2,incl]
x_input = np.concatenate((x_input1,x_input2))
y_input = np.concatenate((y_input1,y_input2))

#15.368591365526559 16.806457045284013 1.6978320432769678 18.372796137399522
#10.095410657922711 8.666100868605136 0.7983823322343429 26.66350227525614
from scipy.optimize import curve_fit
#‘lm’, ‘trf’, ‘dogbox’
yerr = y_input*0 + 0.49/cdelt2/2.355 #error of 1 beam size in sigma
popt, pcov = curve_fit(wind, x_input, y_input, p0=x0,sigma=yerr, absolute_sigma=True, method='lm')
print(popt[0],popt[1],popt[2],popt[3]*180/np.pi)
fit_error = np.sqrt(np.diag(pcov))
print(fit_error[0],fit_error[1],fit_error[2],fit_error[3]*180/np.pi)

r = np.linspace(-10, 10, 200)
v0 = popt[0]
const = popt[1]
incl = popt[3]
pv_v = v0*r*(np.cos(incl) + const*r*np.sin(incl))
pv_p = np.cos(incl)*const*r**2 - r*np.sin(incl)
pv_v_red = ((vlsr + pv_v)*1000-crval1)/cdelt1
pv_p_red = (-pv_p - crval2)/cdelt2 + poff_4A1/cdelt2
pv_v_blue = ((vlsr - pv_v)*1000-crval1)/cdelt1
pv_p_blue = (pv_p - crval2)/cdelt2 + poff_4A1/cdelt2
ax1.plot(pv_v_red, pv_p_red,lw=0.4, linestyle='-', color='k')
ax1.plot(pv_v_blue, pv_p_blue,lw=0.4, linestyle='-', color='k')

v0 = popt[0]
const = popt[2]
incl = popt[3]
pv_v = v0*r*(np.cos(incl) + const*r*np.sin(incl))
pv_p = np.cos(incl)*const*r**2 - r*np.sin(incl)
pv_v_red = ((vlsr + pv_v)*1000-crval1)/cdelt1
pv_p_red = (-pv_p - crval2)/cdelt2 + poff_4A1/cdelt2
pv_v_blue = ((vlsr - pv_v)*1000-crval1)/cdelt1
pv_p_blue = (pv_p - crval2)/cdelt2 + poff_4A1/cdelt2
ax1.plot(pv_v_red, pv_p_red,lw=0.4, linestyle='-', color='b')
ax1.plot(pv_v_blue, pv_p_blue,lw=0.4, linestyle='-', color='b')
##########

#p19 = wind(np.array([v19/1000 - vlsr]), v0, const, incl)
#ax1.scatter((v19-crval1)/cdelt1,p19, 15, c='k', marker="+", zorder=50)

levs1 = np.arange(5.,106.,10)*0.005
vmax = np.amax(pvM_rot)
ax1.contourf(pvM_rot,levels=levs1, cmap='Greys',alpha=1,
             aspect=(x_max-x_min)/(y_max-y_min),vmax=1.2*vmax, vmin=-0.2*vmax)
ax1.contour(pvM_rot, levels=levs1,colors='k',linewidths=0.2,aspect=(x_max-x_min)/(y_max-y_min),zorder=30,interpolation='none')

ax1.set_xlim(x_min,x_max)
ax1.set_ylim(y_min,y_max)
plt.xticks(np.arange((-20000-crval1)/cdelt1,(65000-crval1)/cdelt1,20000/cdelt1))
plt.yticks(np.arange((-10-crval2)/cdelt2,(11-crval2)/cdelt2,5/cdelt2))
ax1.set_xticklabels(['-20','0','20','40','60'])
ax1.set_yticklabels(['-10','-5','0','5','10'])     
ax1.text(0.8,0.85, r'Major cut',horizontalalignment='center',verticalalignment='center',transform=ax1.transAxes)


ax0.text(-0.08, 0.5, r'Position offset from the redshifted emission peak (arcsec)',horizontalalignment='center',verticalalignment='center',\
         transform=ax0.transAxes,fontsize=10,rotation='vertical')
ax0.text(0.25, -0.07, r'V$_{LSR}$ (km s$^{-1}$)',horizontalalignment='center',verticalalignment='center',\
         transform=ax0.transAxes,fontsize=10)

plt.savefig('pv_red_2fits_6.86.pdf')
```
